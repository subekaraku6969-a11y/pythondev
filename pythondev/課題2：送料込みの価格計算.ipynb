```python
price = int(input("税抜価格を入力してください："))
tax_included = int(round(price * 1.1)) 
print(f"税込価格は{tax_included}円です。")
if tax_included>= 2000:
   print("送料は無料です")
   print(f"送料込みの価格は{tax_included}円です。") 
else:
   print("送料として350円かかります")
   tax_included+= 350
   print(f"送料込みの価格は{tax_included}円です。")


```
