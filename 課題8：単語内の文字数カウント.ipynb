```python
words = []
char_count = {chr(c): 0 for c in range(ord('a'), ord('z') + 1)}
while True:
    word = input("英単語を入力してください：").strip().lower()
    if word == "":
        break
    words.append(word)
    for char in word:
        if char in char_count:
            char_count[char] += 1
sorted_words = sorted(words)
print(f"入力した英単語：{sorted_words}")
for char in sorted(char_count.keys()):
    count = char_count[char]
    if count > 0:
        print(f"{char}が{count}個ありました")            
```
